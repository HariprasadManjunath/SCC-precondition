```python
# ================================================================
# SYNTHETIC DIRECTED GRAPH EXPERIMENT
#
# Synthetic directed graph numerical experiment
# Numerical experiment: n=10,000; 200 SCCs; 50 vertices/SCC;
# condensation DAG depth 199; 2 weak + 2 moderate instances.
#
#
#
# Main reported methods:
#   Unpreconditioned GMRES, Jacobi, SCC block Jacobi,
#   proposed SCC-aware ILU_r (bandwidth-based), ILU(0)-type, ILUTP(1e-3, fill=10), AIR-AMG(one-point),
#   global SCC-Neumann q=3,4,
#   band SCC-Neumann (rho,q)=(2,3),(2,4),(3,3),(3,4),(4,4),(4,8).
#
# Proposed SCC-aware ILU_r:
#   Keep every within-SCC entry. For entries between different SCCs,
#   retain only 0 <= column-row <= r in SCC-topological scalar ordering.
#   Factor the truncated matrix with zero-drop ILU and right-precondition
#   the original, untruncated system.
#
# All solves use RIGHT preconditioning and all final residuals use original A.
# ================================================================

import sys, subprocess
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'numpy', 'scipy', 'pandas', 'networkx', 'matplotlib', 'pyamg'])

import os, time, gc, warnings
import numpy as np
import pandas as pd
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import networkx as nx
import matplotlib.pyplot as plt

from scipy.sparse.csgraph import connected_components
from scipy.sparse.linalg import LinearOperator, gmres, spilu

warnings.filterwarnings("ignore")

# ================================================================
# 1. EXPERIMENT PARAMETERS AND EXPLICIT GENERATOR ASSUMPTIONS
# ================================================================

N = 10_000
NUM_SCCS = 200
SCC_SIZE = 50
assert NUM_SCCS * SCC_SIZE == N

# Two independent instances per coupling regime.
INSTANCES_PER_REGIME = 2

# The experiment does not give numerical definitions of weak/moderate coupling.
# Reproducible operational choices:
# Exact condensation-DAG arc counts for the four requested instances.
# The mandatory 199-edge chain guarantees DAG depth 199.
TARGET_DAG_EDGES = {
    ("Weak", 1): 199,
    ("Weak", 2): 199,
    ("Moderate", 1): 576,
    ("Moderate", 2): 599,
}

# Preserve the attached generator's cycle + random internal-edge model.
INTERNAL_EDGE_PROB = 0.01
MAX_SCC_FORWARD_DISTANCE = 10
GRAPH_BASE_SEED = 42

# GMRES tolerance.
GMRES_RTOL = 1e-10
GMRES_ATOL = 0.0
GMRES_RESTART = 50
GMRES_MAXITER = 500

# Neumann parameter combinations.
GLOBAL_Q_VALUES = [3, 4]
BAND_NEUMANN_PARAMETERS = [
    (2, 3), (2, 4), (3, 3), (3, 4), (4, 4), (4, 8)
]

# Scalar bandwidths for the proposed SCC-aware ILU_r method.
ILUR_BANDWIDTHS = [500, 1000, 1500, 2000, 3000, 4000]

# ILUTP configuration.
ILUTP_DROP_TOL = 1e-3
ILUTP_FILL_FACTOR = 10.0

# Optional AIR-AMG.
RUN_AIR = True

RHS_BASE_SEED = 12345
OUTPUT_DIR = "/content/synthetic_scc_experiment"
os.makedirs(OUTPUT_DIR, exist_ok=True)

RESULTS_CSV = os.path.join(OUTPUT_DIR, "synthetic_results_by_instance.csv")
AVERAGES_CSV = os.path.join(OUTPUT_DIR, "synthetic_results_averaged.csv")
GRAPH_CSV = os.path.join(OUTPUT_DIR, "synthetic_graph_statistics.csv")

print("Synthetic directed graph experiment")
print(f"n={N}, SCCs={NUM_SCCS}, SCC size={SCC_SIZE}")
print(f"GMRES rtol={GMRES_RTOL}, restart={GMRES_RESTART}, maxiter={GMRES_MAXITER}")
print("Global q:", GLOBAL_Q_VALUES)
print("Band (rho,q):", BAND_NEUMANN_PARAMETERS)
print("Exact DAG edge targets:", TARGET_DAG_EDGES)

# ================================================================
# 2. GRAPH GENERATOR
# ================================================================

def generate_synthetic_graph(regime, instance, seed):
    """
    Construct 200 SCCs of exactly 50 vertices.

    Inside each SCC:
      - one directed cycle (guarantees strong connectivity);
      - random directed internal edges with probability 0.01.

    Between SCCs:
      - all edges point from a lower-index SCC to a higher-index SCC;
      - a mandatory edge k -> k+1 creates a chain of length 199;
      - additional distinct SCC-to-SCC arcs are sampled without replacement
        to meet the exact per-instance condensation-DAG edge target.

    The benchmark specification gives the SCC sizes and DAG depth, but not this precise
    random graph-generation recipe or the coupling counts. The weak/moderate
    settings above are therefore declared modelling assumptions.
    """
    rng = np.random.default_rng(seed)
    rows, cols = [], []

    # Internal SCC edges.
    for k in range(NUM_SCCS):
        start = k * SCC_SIZE
        v = np.arange(start, start + SCC_SIZE)

        # Directed cycle.
        rows.append(v)
        cols.append(np.roll(v, -1))

        # Bernoulli random internal edges; diagonal excluded.
        mask = rng.random((SCC_SIZE, SCC_SIZE)) < INTERNAL_EDGE_PROB
        np.fill_diagonal(mask, False)
        rr, cc = np.nonzero(mask)
        rows.append(start + rr)
        cols.append(start + cc)

    # Inter-SCC edges: guaranteed chain plus additional forward links.
    target_dag_edges = TARGET_DAG_EDGES[(regime, instance)]

    # First add exactly one edge for every chain arc k -> k+1.
    selected_pairs = [(k, k + 1) for k in range(NUM_SCCS - 1)]

    # Add distinct forward SCC arcs until the requested DAG edge count is met.
    # Sampling is over SCC pairs, not vertex pairs, so the condensation graph
    # has exactly the requested number of distinct arcs.
    candidates = [
        (k, j)
        for k in range(NUM_SCCS - 1)
        for j in range(k + 2, min(NUM_SCCS, k + MAX_SCC_FORWARD_DISTANCE + 1))
    ]
    extra_needed = target_dag_edges - (NUM_SCCS - 1)
    if extra_needed < 0 or extra_needed > len(candidates):
        raise ValueError("Requested DAG edge count is infeasible.")
    chosen_ids = rng.choice(len(candidates), size=extra_needed, replace=False)
    selected_pairs.extend(candidates[int(i)] for i in chosen_ids)

    # Realize each selected SCC arc by one randomly selected vertex edge.
    for source_component, target_component in selected_pairs:
        src_start = source_component * SCC_SIZE
        u = int(rng.integers(src_start, src_start + SCC_SIZE))
        v = int(rng.integers(
            target_component * SCC_SIZE,
            (target_component + 1) * SCC_SIZE
        ))
        rows.append(np.array([u]))
        cols.append(np.array([v]))

    rr = np.concatenate(rows).astype(np.int64)
    cc = np.concatenate(cols).astype(np.int64)

    W = sp.csr_matrix(
        (np.ones(len(rr), dtype=np.float64), (rr, cc)),
        shape=(N, N)
    )
    W.sum_duplicates()
    W.data[:] = 1.0
    W.eliminate_zeros()

    return W

# ================================================================
# 3. SCC DECOMPOSITION, DAG, TOPOLOGICAL ORDERING
# ================================================================

def get_scc_ordering(W):
    ncomp, labels = connected_components(
        W, directed=True, connection="strong", return_labels=True
    )
    sizes = np.bincount(labels, minlength=ncomp)

    if ncomp != NUM_SCCS:
        raise RuntimeError(f"Expected {NUM_SCCS} SCCs, detected {ncomp}.")
    if not np.all(sizes == SCC_SIZE):
        raise RuntimeError("Detected SCC sizes are not all exactly 50.")

    coo = W.tocoo()
    cross = labels[coo.row] != labels[coo.col]
    pairs = np.unique(
        np.column_stack((labels[coo.row[cross]], labels[coo.col[cross]])),
        axis=0
    )

    G = nx.DiGraph()
    G.add_nodes_from(range(ncomp))
    G.add_edges_from(pairs.tolist())

    if not nx.is_directed_acyclic_graph(G):
        raise RuntimeError("Condensation graph is not acyclic.")

    topo = list(nx.topological_sort(G))

    # Longest path measured in edges.
    dag_depth = nx.dag_longest_path_length(G)
    if dag_depth != NUM_SCCS - 1:
        raise RuntimeError(
            f"Expected condensation depth 199, obtained {dag_depth}."
        )

    # Reorder vertices by topological SCC order.
    parts = [np.flatnonzero(labels == c) for c in topo]
    perm = np.concatenate(parts)
    labels_ordered = labels[perm]

    blocks = [
        np.arange(k * SCC_SIZE, (k + 1) * SCC_SIZE, dtype=np.int64)
        for k in range(NUM_SCCS)
    ]

    return {
        "n_scc": ncomp,
        "labels": labels,
        "labels_ordered": labels_ordered,
        "sizes": sizes,
        "topo": topo,
        "perm": perm,
        "blocks": blocks,
        "dag": G,
        "dag_depth": dag_depth,
        "dag_edges": G.number_of_edges()
    }

# ================================================================
# 4. BUILD PERMUTED SHIFTED LAPLACIAN A = D - E
# ================================================================

def build_system(W_original, info):
    perm = info["perm"]
    labels_ord = info["labels_ordered"]

    W = W_original[perm, :][:, perm].tocsr()
    W.sum_duplicates()
    W.eliminate_zeros()

    outdegree = np.asarray(W.sum(axis=1)).ravel()
    A = (
        sp.eye(N, format="csr")
        + sp.diags(outdegree, format="csr")
        - W
    ).tocsr()
    A.eliminate_zeros()

    coo = W.tocoo()
    cross = labels_ord[coo.row] != labels_ord[coo.col]
    E = sp.coo_matrix(
        (coo.data[cross], (coo.row[cross], coo.col[cross])),
        shape=(N, N)
    ).tocsr()
    E.eliminate_zeros()

    # D retains the ORIGINAL out-degree diagonal.
    D = (A + E).tocsr()
    D.eliminate_zeros()

    if (A - (D - E)).nnz != 0:
        raise RuntimeError("A = D - E identity failed.")

    return A, D, E, W, info["blocks"], labels_ord

# ================================================================
# 5. SCC BLOCK FACTORIZATION AND D^{-1} APPLICATION
# ================================================================

def factorize_D(D, blocks):
    t0 = time.perf_counter()
    factors = []
    for idx in blocks:
        block = D[idx, :][:, idx].tocsc()
        factors.append(spla.splu(block))
    return factors, time.perf_counter() - t0

def make_Dsolve(factors, blocks):
    def apply(v):
        v = np.asarray(v).reshape(-1)
        out = np.zeros(N, dtype=np.float64)
        for idx, lu in zip(blocks, factors):
            out[idx] = lu.solve(v[idx])
        return out
    return apply

# ================================================================
# 6. CONSTRUCT T = E D^{-1} AND SCC-LEVEL T_rho
# ================================================================

def build_T( E, factors, blocks, row_batch=256 ):
    """
    Construct T = E D^{-1} block-column by block-column.

    For target SCC j:
       T[:, S_j] = E[:, S_j] D_j^{-1}.
    The multiplication is performed through transposed LU solves.
    """
    t0 = time.perf_counter()
    E = E.tocsr()
    row_parts, col_parts, data_parts = [], [], []

    row_batches = [
        np.arange(i, min(i + row_batch, N), dtype=np.int64)
        for i in range(0, N, row_batch)
    ]

    for block_id, vertices in enumerate(blocks):
        lu = factors[block_id]
        E_block = E[:, vertices].tocsr()

        if E_block.nnz == 0:
            continue

        for ridx in row_batches:
            B = E_block[ridx, :].toarray()
            if not np.any(B):
                continue

            # X = B D_j^{-1}; hence X^T = D_j^{-T} B^T.
            X = lu.solve(B.T, trans="T").T
            ii, jj = np.nonzero(X)

            if len(ii):
                row_parts.append(ridx[ii])
                col_parts.append(vertices[jj])
                data_parts.append(X[ii, jj])

    if data_parts:
        T = sp.coo_matrix(
            (
                np.concatenate(data_parts),
                (np.concatenate(row_parts), np.concatenate(col_parts))
            ),
            shape=(N, N)
        ).tocsr()
    else:
        T = sp.csr_matrix((N, N), dtype=np.float64)

    T.sum_duplicates()
    T.eliminate_zeros()
    return T, time.perf_counter() - t0

def truncate_T_by_scc(T, rho):
    """
    Keep SCC block superdiagonals 1,...,rho.
    This is SCC-level truncation, not scalar bandwidth truncation.
    """
    coo = T.tocoo()
    row_block = coo.row // SCC_SIZE
    col_block = coo.col // SCC_SIZE
    offset = col_block - row_block

    keep = (offset >= 1) & (offset <= rho)

    T_rho = sp.coo_matrix(
        (coo.data[keep], (coo.row[keep], coo.col[keep])),
        shape=T.shape
    ).tocsr()
    T_rho.sum_duplicates()
    T_rho.eliminate_zeros()
    return T_rho

# ================================================================
# 7. PRECONDITIONER OPERATORS
# ================================================================

def make_neumann_operator(Dsolve, Tmat, q):
    """
    M_(q) = D^{-1} sum_{j=0}^q T^j.
    Applies the polynomial to v, then applies D^{-1}.
    """
    def matvec(v):
        z = np.asarray(v).copy()
        total = z.copy()
        for _ in range(q):
            z = Tmat @ z
            total += z
        return Dsolve(total)

    return LinearOperator((N, N), matvec=matvec, dtype=np.float64)

def make_global_neumann_operator(Dsolve, E, q):
    """
    Matrix-free T action: T z = E D^{-1} z.
    Equivalent to D^{-1} sum_{j=0}^q T^j.
    """
    def matvec(v):
        z = np.asarray(v).copy()
        total = z.copy()
        for _ in range(q):
            z = E @ Dsolve(z)
            total += z
        return Dsolve(total)

    return LinearOperator((N, N), matvec=matvec, dtype=np.float64)

# ================================================================
# 8. RIGHT-PRECONDITIONED GMRES AND TIMING
# ================================================================

def solve_one(method, A, b, M, setup_time, metadata=None):
    """
    Solve A M y = b, recover x=M y.
    The reported residual is ||A x-b||_2 / ||b||_2 using original A.
    """
    AM = LinearOperator(
        (N, N),
        matvec=lambda y: A @ M.matvec(y),
        dtype=np.float64
    )

    # Average preconditioner application time (warm-up + timed calls).
    probe = np.random.default_rng(9876).standard_normal(N)
    M.matvec(probe)
    repeats = 5
    t0 = time.perf_counter()
    for _ in range(repeats):
        M.matvec(probe)
    apply_time = (time.perf_counter() - t0) / repeats

    count = [0]
    def callback(_):
        count[0] += 1

    t0 = time.perf_counter()
    y, info = gmres(
        AM, b,
        rtol=GMRES_RTOL,
        atol=GMRES_ATOL,
        restart=GMRES_RESTART,
        maxiter=GMRES_MAXITER,
        callback=callback,
        callback_type="pr_norm"
    )
    solve_time = time.perf_counter() - t0

    x = M.matvec(y)
    residual = np.linalg.norm(A @ x - b) / np.linalg.norm(b)

    rec = {
        "Method": method,
        "Setup_Time": setup_time,
        "Apply_Time": apply_time,
        "Solve_Time": solve_time,
        "Iterations": count[0],
        "GMRES_Info": int(info),
        "True_Residual": residual
    }
    if metadata:
        rec.update(metadata)

    print(
        f"{method:38s} iter={count[0]:4d} "
        f"res={residual:.3e} solve={solve_time:.5f}s "
        f"apply={apply_time:.3e}s info={info}",
        flush=True
    )
    return rec

# ================================================================
# 9. PROPOSED SCC-AWARE BANDWIDTH-BASED ILU_r
# ================================================================

def build_ilu_r_matrix(A, labels_ordered, bandwidth):
    """Keep all within-SCC entries; between SCCs keep only
    0 <= column-row <= bandwidth in SCC-topological scalar ordering.
    """
    coo = A.tocoo()
    same_scc = labels_ordered[coo.row] == labels_ordered[coo.col]
    offset = coo.col - coo.row
    in_forward_band = (offset >= 0) & (offset <= bandwidth)
    keep = same_scc | ((~same_scc) & in_forward_band)

    A_r = sp.coo_matrix(
        (coo.data[keep], (coo.row[keep], coo.col[keep])),
        shape=A.shape
    ).tocsr()
    A_r.sum_duplicates()
    A_r.eliminate_zeros()
    return A_r


def build_ilu_r_preconditioner(A, labels_ordered, bandwidth):
    """Factor truncated A_r with zero-drop ILU; use as right
    preconditioner for the original, untruncated A.
    """
    t0 = time.perf_counter()
    A_r = build_ilu_r_matrix(A, labels_ordered, bandwidth)
    truncation_time = time.perf_counter() - t0

    t0 = time.perf_counter()
    ilu_factor = spilu(
        A_r.tocsc(),
        drop_tol=0.0,
        fill_factor=1.0,
        permc_spec="NATURAL",
        diag_pivot_thresh=0.0
    )
    factor_time = time.perf_counter() - t0

    Mop = LinearOperator(
        A.shape, matvec=ilu_factor.solve, dtype=np.float64
    )
    metadata = {
        "Bandwidth": int(bandwidth),
        "A_r_NNZ": int(A_r.nnz),
        "Factor_NNZ": int(ilu_factor.L.nnz + ilu_factor.U.nnz),
        "Truncation_Time": truncation_time,
        "Factorization_Time": factor_time
    }
    return Mop, ilu_factor, A_r, truncation_time + factor_time, metadata


# ================================================================
# 10. RUN ONE GRAPH INSTANCE
# ================================================================

all_records = []
graph_records = []

def run_instance(regime, instance, seed):
    print("\n" + "=" * 80)
    print(f"RUN: {regime} coupling | instance {instance} | seed {seed}")
    print("=" * 80, flush=True)

    # Generate graph and verify SCC structure.
    W0 = generate_synthetic_graph(regime, instance, seed)
    info = get_scc_ordering(W0)

    print(
        f"Graph: n={N}, m={W0.nnz}, SCCs={info['n_scc']}, "
        f"SCC size={SCC_SIZE}, DAG edges={info['dag_edges']}, "
        f"DAG depth={info['dag_depth']}",
        flush=True
    )

    A, D, E, W, blocks, labels_ord = build_system(W0, info)

    # Factorize D and construct D^{-1}.
    factors, D_factor_time = factorize_D(D, blocks)
    Dsolve = make_Dsolve(factors, blocks)

    # Verify D solve.
    test = np.random.default_rng(seed + 700).standard_normal(N)
    dsolve_err = np.linalg.norm(D @ Dsolve(test) - test) / np.linalg.norm(test)
    if dsolve_err > 1e-9:
        raise RuntimeError(f"D solve check failed: {dsolve_err}")

    # Construct T=E D^{-1}; this is reused for all Neumann variants.
    T, T_setup_time = build_T(E, factors, blocks)

    # Verify strict block upper triangularity.
    tcoo = T.tocoo()
    if np.any((tcoo.col // SCC_SIZE) <= (tcoo.row // SCC_SIZE)):
        raise RuntimeError("T is not strictly block upper triangular.")

    # Same RHS for all methods on this instance.
    rhs_rng = np.random.default_rng(RHS_BASE_SEED + seed)
    b = rhs_rng.standard_normal(N)

    # Graph statistics.
    graph_records.append({
        "Regime": regime,
        "Instance": instance,
        "Seed": seed,
        "Vertices": N,
        "Edges": W0.nnz,
        "SCC_Count": info["n_scc"],
        "SCC_Size_Min": int(info["sizes"].min()),
        "SCC_Size_Max": int(info["sizes"].max()),
        "DAG_Edges": info["dag_edges"],
        "Target_DAG_Edges": TARGET_DAG_EDGES[(regime, instance)],
        "DAG_Depth": info["dag_depth"],
        "nnz_A": A.nnz,
        "nnz_D": D.nnz,
        "nnz_E": E.nnz,
        "nnz_T": T.nnz,
        "D_Factor_Time": D_factor_time,
        "T_Setup_Time": T_setup_time,
        "D_Solve_Check": dsolve_err
    })

    local = []

    def add(record):
        record["Regime"] = regime
        record["Instance"] = instance
        record["Seed"] = seed
        local.append(record)
        all_records.append(record)

    # 1. Unpreconditioned GMRES.
    Iop = LinearOperator((N, N), matvec=lambda x: x, dtype=np.float64)
    add(solve_one("Unpreconditioned GMRES", A, b, Iop, 0.0))

    # 2. Scalar Jacobi.
    dinv = 1.0 / A.diagonal()
    Jop = LinearOperator((N, N), matvec=lambda x: dinv * x, dtype=np.float64)
    add(solve_one("Jacobi", A, b, Jop, 0.0))

    # 3. SCC block Jacobi (q=0).
    Bop = LinearOperator((N, N), matvec=Dsolve, dtype=np.float64)
    add(solve_one("SCC Block Jacobi", A, b, Bop, D_factor_time))

    # 4. Proposed SCC-aware bandwidth-based ILU_r.
    # All within-SCC entries are retained. The scalar upper bandwidth is
    # applied only to entries connecting different SCCs.
    for bandwidth in ILUR_BANDWIDTHS:
        print(f"Building proposed SCC-aware ILU_r, bandwidth={bandwidth}...", flush=True)
        Mop, ilu_r_factor, A_r, ilu_r_setup, ilu_r_meta = (
            build_ilu_r_preconditioner(A, labels_ord, bandwidth)
        )
        add(solve_one(
            f"Proposed SCC-aware ILU_r (bandwidth={bandwidth})",
            A, b, Mop, ilu_r_setup, ilu_r_meta
        ))
        del Mop, ilu_r_factor, A_r
        gc.collect()

    # 5. Global SCC-Neumann q=3,4.
    for q in GLOBAL_Q_VALUES:
        Mop = make_global_neumann_operator(Dsolve, E, q)
        add(solve_one(
            f"Global SCC-Neumann q={q}", A, b, Mop,
            D_factor_time + T_setup_time,
            {"q": q, "T_nnz": T.nnz}
        ))

    # 6. Band-truncated SCC-Neumann exact parameter pairs.
    for rho, q in BAND_NEUMANN_PARAMETERS:
        Trho = truncate_T_by_scc(T, rho)
        Mop = make_neumann_operator(Dsolve, Trho, q)
        add(solve_one(
            f"Band SCC-Neumann rho={rho}, q={q}",
            A, b, Mop,
            D_factor_time + T_setup_time,
            {"rho": rho, "q": q, "T_rho_nnz": Trho.nnz}
        ))
        del Trho

    # 7. ILU(0)-type baseline.
    print("Building ILU(0)-type...", flush=True)
    t0 = time.perf_counter()
    ilu0 = spilu(
        A.tocsc(),
        drop_tol=0.0,
        fill_factor=1.0,
        permc_spec="NATURAL",
        diag_pivot_thresh=0.0
    )
    ilu0_setup = time.perf_counter() - t0
    ilu0op = LinearOperator((N, N), matvec=ilu0.solve, dtype=np.float64)
    add(solve_one(
        "ILU(0)-type", A, b, ilu0op, ilu0_setup,
        {"Factor_NNZ": ilu0.L.nnz + ilu0.U.nnz}
    ))
    del ilu0, ilu0op
    gc.collect()

    # 8. ILUTP (drop=1e-3, fill=10).
    print("Building ILUTP (drop=1e-3, fill=10)...", flush=True)
    t0 = time.perf_counter()
    ilutp = spilu(
        A.tocsc(),
        drop_tol=ILUTP_DROP_TOL,
        fill_factor=ILUTP_FILL_FACTOR,
        permc_spec="COLAMD"
    )
    ilutp_setup = time.perf_counter() - t0
    ilutpop = LinearOperator((N, N), matvec=ilutp.solve, dtype=np.float64)
    add(solve_one(
        "ILUTP (1e-3, fill=10)", A, b, ilutpop, ilutp_setup,
        {"Factor_NNZ": ilutp.L.nnz + ilutp.U.nnz}
    ))
    del ilutp, ilutpop
    gc.collect()

    # 9. AIR-AMG one-point (included in the numerical comparison).
    if RUN_AIR:
        try:
            import pyamg
            print("Building AIR-AMG one-point...", flush=True)
            t0 = time.perf_counter()
            air = pyamg.air_solver(A, interpolation="one_point")
            air_setup = time.perf_counter() - t0
            airop = LinearOperator(
                (N, N),
                matvec=lambda x: air.solve(
                    x, tol=0.0, maxiter=1, cycle="V"
                ),
                dtype=np.float64
            )
            add(solve_one(
                "AIR-AMG (one-point)", A, b, airop, air_setup
            ))
            del air, airop
            gc.collect()
        except Exception as exc:
            print("AIR-AMG unavailable for this instance:", repr(exc), flush=True)
            add({
                "Method": "AIR-AMG (one-point)",
                "Setup_Time": np.nan,
                "Apply_Time": np.nan,
                "Solve_Time": np.nan,
                "Iterations": np.nan,
                "GMRES_Info": -999,
                "True_Residual": np.nan,
                "Status": f"Unavailable: {exc}"
            })

    # Checkpoint after every instance.
    pd.DataFrame(all_records).to_csv(RESULTS_CSV, index=False)
    pd.DataFrame(graph_records).to_csv(GRAPH_CSV, index=False)

    print(f"Checkpoint saved after {regime} instance {instance}.", flush=True)

    del W0, A, D, E, W, T, factors
    gc.collect()

# ================================================================
# 11. FOUR INSTANCES: 2 WEAK + 2 MODERATE
# ================================================================

instance_id = 0
for regime in ["Weak", "Moderate"]:
    for local_instance in range(1, INSTANCES_PER_REGIME + 1):
        instance_id += 1
        seed = GRAPH_BASE_SEED + 1000 * instance_id
        run_instance(regime, local_instance, seed)

# ================================================================
# 12. AGGREGATE ARITHMETIC MEANS OVER FOUR INSTANCES
# ================================================================

results_df = pd.DataFrame(all_records)
graph_df = pd.DataFrame(graph_records)

# Average only numerical metrics. A method unavailable in any instance
# will show NaN for the affected aggregate.
metric_cols = [
    "Iterations", "Setup_Time", "Apply_Time",
    "Solve_Time", "True_Residual"
]
averages_df = (
    results_df.groupby("Method", dropna=False)[metric_cols]
    .mean()
    .reset_index()
)

results_df.to_csv(RESULTS_CSV, index=False)
averages_df.to_csv(AVERAGES_CSV, index=False)
graph_df.to_csv(GRAPH_CSV, index=False)

print("\n" + "=" * 80)
print("GRAPH STRUCTURE CHECKS")
print("=" * 80)
display(graph_df)

assert len(graph_df) == 4
assert (graph_df["Vertices"] == 10000).all()
assert (graph_df["SCC_Count"] == 200).all()
assert (graph_df["SCC_Size_Min"] == 50).all()
assert (graph_df["SCC_Size_Max"] == 50).all()
assert (graph_df["DAG_Depth"] == 199).all()
assert (graph_df["DAG_Edges"] == graph_df["Target_DAG_Edges"]).all()

print("All four graph instances passed the structural checks, including exact DAG edge counts.")

print("\n" + "=" * 80)
print("AVERAGED RESULTS — ARITHMETIC MEAN OVER FOUR INSTANCES")
print("=" * 80)
display(averages_df)

print("\nPer-instance results:")
display(results_df)

# ================================================================
# 13. PLOTS — ONE BELOW ANOTHER
# ================================================================

plot_df = averages_df.copy()

plt.figure(figsize=(15, 5))
plt.bar(plot_df["Method"], plot_df["Iterations"])
plt.ylabel("Average GMRES iterations")
plt.title("Synthetic experiment: average GMRES iterations")
plt.xticks(rotation=75, ha="right")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "average_iterations.png"), dpi=200)
plt.show()

plt.figure(figsize=(15, 5))
plt.bar(plot_df["Method"], plot_df["Solve_Time"])
plt.ylabel("Average solve time (seconds)")
plt.title("Synthetic experiment: average GMRES solve time")
plt.xticks(rotation=75, ha="right")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "average_solve_time.png"), dpi=200)
plt.show()

plt.figure(figsize=(15, 5))
plt.bar(plot_df["Method"], plot_df["Apply_Time"])
plt.ylabel("Average preconditioner application time (seconds)")
plt.title("Synthetic experiment: average preconditioner application time")
plt.xticks(rotation=75, ha="right")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "average_apply_time.png"), dpi=200)
plt.show()

plt.figure(figsize=(15, 5))
plt.semilogy(
    np.arange(len(plot_df)),
    plot_df["True_Residual"],
    marker="o"
)
plt.xticks(np.arange(len(plot_df)), plot_df["Method"], rotation=75, ha="right")
plt.ylabel("Average true relative residual")
plt.title("Synthetic experiment: true residual against original A")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "average_true_residual.png"), dpi=200)
plt.show()

print("\nExperiment finished.")
print("Results:", RESULTS_CSV)
print("Averages:", AVERAGES_CSV)
print("Graph checks:", GRAPH_CSV)
print("Plots and CSV files are in:", OUTPUT_DIR)

# Optional: download result files from Colab.
# from google.colab import files
# files.download(AVERAGES_CSV)
```
